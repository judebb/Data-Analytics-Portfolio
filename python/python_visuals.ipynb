## Monthly Sales Volume Over Time

```python
sales['Sale_Date'] = pd.to_datetime(sales['Sale_Date'])  #edit date format

monthly_sales = sales.set_index('Sale_Date').resample('M')['Sales_Amount'].sum() #set sale date as index, organize by Month

plt.figure(figsize=(10, 5))
monthly_sales.plot(marker='o')
plt.title("Monthly Sales Volume Over Time")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.tight_layout()
plt.show()
```

## Avg Delivery Time Per Month

```python
sales_data['Order_Date'] = pd.to_datetime(sales_data['Order_Date'])
sales_data['Delivery_Date'] = pd.to_datetime(sales_data['Delivery_Date'])
sales_data['Delivery_Days'] = (sales_data['Delivery_Date'] - sales_data['Order_Date']).dt.days

monthly_delivery = sales_data.set_index('Order_Date').resample('M')['Delivery_Days'].mean()

plt.figure(figsize=(10, 5))
monthly_delivery.plot(marker='o', color='darkorange') ##what THISDO
plt.title("Average Delivery Time Per Month")
plt.xlabel("Month")
plt.ylabel("Avg Delivery Days")
plt.tight_layout()
plt.show()
```
