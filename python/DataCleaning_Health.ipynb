```python
# standard date format
data['Date of Admission'] = pd.to_datetime(data['Date of Admission'], errors='coerce')
data['Discharge Date']    = pd.to_datetime(data['Discharge Date'],    errors='coerce')

data
```

## Adds columns for admission month, Qtr, and length of stay

```python
#FEATURE ENGINEERING
data['admission_year'] = data['Date of Admission'].dt.year     #datetime functions:
data['admission_month_name'] = data['Date of Admission'].dt.month_name()
data['admission_quarter'] = data['Date of Admission'].dt.to_period('Q')

data['length_of_stay(days)'] = (data['Discharge Date'] - data['Date of Admission']).dt.days

data["Billing Amount"] = data["Billing Amount"].round(1)

data.head(10)
```
